# Praktikum Pengolahan Citra Digital
## Modul 2 – Operasi dan Ruang Warna

Ekstraksi objek warna menggunakan ruang warna YCbCr dengan OpenCV.

### Import Library

```python
import cv2 as cv
import numpy as np
```

### Membaca Citra

```python
img = cv.imread("buah.png")
```

### Konversi Ruang Warna BGR ke YCbCr

```python
ycbcr = cv.cvtColor(img, cv.COLOR_BGR2YCrCb)
```

### Threshold Warna Merah / Coklat

```python
lower_color = np.array([0, 150, 80])
upper_color = np.array([255, 200, 130])

mask = cv.inRange(ycbcr, lower_color, upper_color)
```

### Ekstraksi Objek Berwarna

```python
result = cv.bitwise_and(img, img, mask=mask)
```

### Menampilkan Hasil

```python
cv.namedWindow("Citra Asli", cv.WINDOW_NORMAL)
cv.namedWindow("Mask", cv.WINDOW_NORMAL)
cv.namedWindow("Hasil Ekstraksi", cv.WINDOW_NORMAL)

cv.imshow("Citra Asli", img)
cv.imshow("Mask", mask)
cv.imshow("Hasil Ekstraksi", result)

cv.waitKey(0)
cv.destroyAllWindows()
```
