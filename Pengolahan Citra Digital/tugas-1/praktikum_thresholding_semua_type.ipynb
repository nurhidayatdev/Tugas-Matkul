## Image Thresholding Manual (Semua Type)

Notebook ini menampilkan **semua tipe thresholding (1–5)** karena sebelumnya hanya satu tipe yang dipanggil.

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

### Fungsi my_threshold

```python
def my_threshold(img, thresh, max_val, type_threshold):
    result = np.zeros_like(img)

    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            pixel = img[i, j]

            if type_threshold == 1:  # Binary
                result[i, j] = max_val if pixel > thresh else 0

            elif type_threshold == 2:  # Binary Inverted
                result[i, j] = 0 if pixel > thresh else max_val

            elif type_threshold == 3:  # Truncate
                result[i, j] = thresh if pixel > thresh else pixel

            elif type_threshold == 4:  # Truncate to Zero
                result[i, j] = pixel if pixel > thresh else 0

            elif type_threshold == 5:  # Truncate to Zero Inverted
                result[i, j] = 0 if pixel > thresh else pixel

    return result
```

### Menampilkan Semua Type Thresholding

```python
# Baca citra grayscale
img = cv2.imread('low.jpeg', 0)

types = {
    1: 'Binary',
    2: 'Binary Inverted',
    3: 'Truncate',
    4: 'Truncate to Zero',
    5: 'Truncate to Zero Inverted'
}

plt.figure(figsize=(12, 6))
plt.subplot(2, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('Citra Asli')
plt.axis('off')

i = 2
for t, name in types.items():
    hasil = my_threshold(img, 127, 255, t)
    plt.subplot(2, 3, i)
    plt.imshow(hasil, cmap='gray')
    plt.title(name)
    plt.axis('off')
    i += 1

plt.tight_layout()
plt.show()
```
