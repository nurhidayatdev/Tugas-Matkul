## Image Thresholding Manual

Notebook ini berisi implementasi fungsi image thresholding tanpa menggunakan fungsi bawaan OpenCV.

```python
import cv2 as cv
import numpy as np
```

### Fungsi my_threshold

```python
def my_threshold(img, thresh, max_val, type_threshold):
    result = np.zeros_like(img)

    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            pixel = img[i, j]

            if type_threshold == 1:  # Binary
                result[i, j] = max_val if pixel > thresh else 0

            elif type_threshold == 2:  # Binary Inverted
                result[i, j] = 0 if pixel > thresh else max_val

            elif type_threshold == 3:  # Truncate
                result[i, j] = thresh if pixel > thresh else pixel

            elif type_threshold == 4:  # Truncate to Zero
                result[i, j] = pixel if pixel > thresh else 0

            elif type_threshold == 5:  # Truncate to Zero Inverted
                result[i, j] = 0 if pixel > thresh else pixel

    return result
```

### Contoh Penggunaan

```python
img = cv.imread('low.jpeg', 0)
```

```python
threshold_types = {
    1: 'Binary',
    2: 'Binary Inverted',
    3: 'Truncate',
    4: 'To Zero',
    5: 'To Zero Inverted'
}

for t, name in threshold_types.items():
    result = my_threshold(img, thresh=127, max_val=255, type_threshold=t)
    cv.imshow(f'Hasil {name}', result)
```
