# Praktikum Pengolahan Citra Digital

Operasi Grayscale, Image Addition, dan Image Subtraction menggunakan OpenCV

```python
import cv2 as cv
import numpy as np
```

## Membaca Gambar dan Konversi ke Grayscale

```python
img_soekarno = cv.imread("soekarno.jpeg")
img_hatta = cv.imread("hatta.jpeg")

gray_soekarno = cv.cvtColor(img_soekarno, cv.COLOR_BGR2GRAY)
gray_hatta = cv.cvtColor(img_hatta, cv.COLOR_BGR2GRAY)
```

## Menyesuaikan Ukuran Gambar

```python
h1, w1 = gray_soekarno.shape
h2, w2 = gray_hatta.shape

min_h = min(h1, h2)
min_w = min(w1, w2)

gray_soekarno = gray_soekarno[0:min_h, 0:min_w]
gray_hatta = gray_hatta[0:min_h, 0:min_w]
```

## Image Addition dan Image Subtraction

```python
image_addition = cv.add(gray_soekarno, gray_hatta)
image_subtraction = cv.subtract(gray_soekarno, gray_hatta)
```

## Menampilkan Hasil

```python
cv.namedWindow("Grayscale Soekarno", cv.WINDOW_NORMAL)
cv.namedWindow("Grayscale Hatta", cv.WINDOW_NORMAL)
cv.namedWindow("Image Addition", cv.WINDOW_NORMAL)
cv.namedWindow("Image Subtraction", cv.WINDOW_NORMAL)

cv.imshow("Grayscale Soekarno", gray_soekarno)
cv.imshow("Grayscale Hatta", gray_hatta)
cv.imshow("Image Addition", image_addition)
cv.imshow("Image Subtraction", image_subtraction)

cv.waitKey(0)
cv.destroyAllWindows()
```
