# Histogram Equalization menggunakan OpenCV

Notebook ini melakukan histogram equalization pada gambar `low.jpg` menggunakan fungsi `cv2.equalizeHist()`.

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv.imread('low.jpeg', 0)
equ = cv.equalizeHist(img)
```

```python
res = np.hstack((img, equ))
cv.imwrite('res.jpg', res)
```

```python
plt.figure(figsize=(10,4))
plt.imshow(res, cmap='gray')
plt.axis('off')
plt.title('Original (kiri) vs Equalized (kanan)')
plt.show()
```
