## Histogram Equalization Manual

```python
import cv2 as cv
import numpy as np
```

### Fungsi myEqualizeHist

```python
def myEqualizeHist(img):
    hist = np.zeros(256)
    for pixel in img.flatten():
        hist[pixel] += 1

    cdf = hist.cumsum()
    cdf_normalized = (cdf - cdf.min()) * 255 / (cdf.max() - cdf.min())
    cdf_normalized = cdf_normalized.astype('uint8')

    result = cdf_normalized[img]
    return result
```

```python
img = cv.imread('sutansjahrir.jpg', 0)
hasil = myEqualizeHist(img)
```

### Contoh Penggunaan

```python
cv.imshow('Asli', img)
cv.imshow('Equalized', hasil)
cv.waitKey(0)
cv.destroyAllWindows()
```
