```python
import cv2 as cv
import numpy as np
```

```python
# Membaca gambar dalam mode grayscale
img = cv.imread('jenderalsoedirman.jpeg', 0)
rows, cols = img.shape
```

```python
# 1. Resize: Menggandakan ukuran gambar (fx=2, fy=2) menggunakan interpolasi kubik
img_resize = cv.resize(img, None, fx=-3, fy=-3, interpolation=cv.INTER_CUBIC)
```

```python
# 2. Translation: Menggeser gambar (x=100, y=50)
M = np.float32([[1, 0, 100], [0, 1, 50]])
img_translation = cv.warpAffine(img, M, (cols, rows))
```

```python
# 3. Rotation: Memutar gambar 45 derajat searah jarum jam dengan pusat di tengah
M = cv.getRotationMatrix2D(((cols - 1) / 2.0, (rows - 1) / 2.0), 45, 1)
img_rotation = cv.warpAffine(img, M, (cols, rows))
```

```python
# Menyimpan hasil transformasi ke file baru
cv.imwrite("jenderalsoedirman_resize.jpg", img_resize)
cv.imwrite("jenderalsoedirman_translation.jpg", img_translation)
cv.imwrite("jenderalsoedirman_rotation.jpg", img_rotation)
```

```python
# Menampilkan hasil di jendela baru
cv.imshow('image resize', img_resize)
cv.imshow('image translation', img_translation)
cv.imshow('image rotation', img_rotation)
```
