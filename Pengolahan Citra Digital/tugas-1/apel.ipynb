```python
import numpy as np
import cv2 as cv
from matplotlib import pyplot as plt
```

```python
img = cv.imread('buah.png', cv.IMREAD_COLOR)
```

```python
# Menampilkan histogram untuk channel Biru (Blue)
plt.subplot(1, 3, 1),
plt.hist(img[:, :, 0].ravel(), 256, [0, 256], color='b')
# Menampilkan histogram untuk channel Hijau (Green)
plt.subplot(1, 3, 2),
plt.hist(img[:, :, 1].ravel(), 256, [0, 256], color='g')
# Menampilkan histogram untuk channel Merah (Red)
plt.subplot(1, 3, 3),
plt.hist(img[:, :, 2].ravel(), 256, [0, 256], color='r')
plt.show()
```
