```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt

img = cv.imread('daun.png', 0)

sobelx64f = cv.Sobel(img, cv.CV_64F, 1, 0, ksize=3)
abs_sobelx64f = np.absolute(sobelx64f)
sobel_x = np.uint8(abs_sobelx64f)

sobely64f = cv.Sobel(img, cv.CV_64F, 0, 1, ksize=3)
abs_sobely64f = np.absolute(sobely64f)
sobel_y = np.uint8(abs_sobely64f)

magnitudesobel = cv.magnitude(sobelx64f, sobely64f)
abs_sobel64f = np.absolute(magnitudesobel)
sobel_magnitude = np.uint8(abs_sobel64f)

laplacian64f = cv.Laplacian(img, cv.CV_64F)
abs_laplacian64f = np.absolute(laplacian64f)
laplacian = np.uint8(abs_laplacian64f)

edges_canny = cv.Canny(img, 100, 200)

ret, mask = cv.threshold(img, 200, 255, cv.THRESH_BINARY_INV)

segmentasi_hasil = cv.bitwise_and(img, img, mask=mask)

cv.imshow('1. Citra Asli (Grayscale)', img)
cv.imshow('2. Sobel Horizontal (Gx)', sobel_x)
cv.imshow('3. Sobel Vertikal (Gy)', sobel_y)
cv.imshow('4. Sobel Magnitude', sobel_magnitude)
cv.imshow('5. Laplacian Derivatives', laplacian)
cv.imshow('6. Canny Edge Detection', edges_canny)
cv.imshow('7. Mask Hasil Thresholding', mask)
cv.imshow('8. Daun Hasil Segmentasi', segmentasi_hasil)

cv.waitKey(0)
cv.destroyAllWindows()
```
