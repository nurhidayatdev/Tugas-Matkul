```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt

img = cv.imread('daun.png', 0)

sobelx64f = cv.Sobel(img, cv.CV_64F, 1, 0, ksize=3)
abs_sobelx64f = np.absolute(sobelx64f)
sobel_x = np.uint8(abs_sobelx64f)

sobely64f = cv.Sobel(img, cv.CV_64F, 0, 1, ksize=3)
abs_sobely64f = np.absolute(sobely64f)
sobel_y = np.uint8(abs_sobely64f)

magnitudesobel = cv.magnitude(sobelx64f, sobely64f)
abs_sobel64f = np.absolute(magnitudesobel)
sobel_magnitude = np.uint8(abs_sobel64f)

edges_canny = cv.Canny(img, 100, 200)

cv.imshow('Citra Asli (Grayscale)', img)
cv.imshow('Sobel Magnitude', sobel_magnitude)
cv.imshow('Canny Edge Detection', edges_canny)

cv.waitKey(0)
cv.destroyAllWindows()
```
